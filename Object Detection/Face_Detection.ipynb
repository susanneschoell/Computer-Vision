# Object Detection - Haar Cascade Classifier

### Face Detection

```python
# import libraries
import cv2 as cv
import matplotlib.pyplot as plt

# load the image
img = cv.imread('lena.jpg')
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# apply Laplacian Filter
laplacian = cv.Laplacian(gray, 5)

# display color, gray, laplacian images
plt.figure(figsize=(12,12))
plt.subplot(231)
plt.imshow(img)
plt.subplot(232)
plt.imshow(gray, cmap = 'gray')
plt.subplot(233)
plt.imshow(laplacian, cmap = 'gray')

# load the Haar Cascade Classifier
face_cascade = cv.CascadeClassifier('haarcascade_frontalface_default.xml')

# detect faces in the image
face = face_cascade.detectMultiScale(gray)

# draw rectangles around faces
for (x, y, h, w) in face:
    face = cv.rectangle(img, (x,y), (x+h, y+w), color = (255, 0, 0), thickness = 2)
    face_gray = cv.rectangle(gray, (x,y), (x+h, y+w), color=(255,255,255), thickness = 2)
    laplace = cv.rectangle(laplacian, (x,y), (x+h, y+w), color=(255,255,255), thickness = 2)
    

# display the results
plt.subplot(234)
plt.imshow(face)
plt.subplot(235)
plt.imshow(face_gray, cmap = 'gray')
plt.subplot(236)
plt.imshow(laplace, cmap = 'gray')

plt.show()
```

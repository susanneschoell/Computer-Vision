# Object Detection - Haar Cascade Classifier

### Face Detection

```python
# import libraries
import cv2 as cv
import matplotlib.pyplot as plt

# load the image
img = cv.imread('lena.jpg')
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# apply Laplacian Filter
laplacian = cv.Laplacian(gray, 5)

# display color, gray, laplacian images
plt.figure(figsize=(12,12))
plt.subplot(221)
plt.imshow(img)
plt.subplot(222)
plt.imshow(gray, cmap = 'gray')

# load the Haar Cascade Classifier
face_cascade = cv.CascadeClassifier('haarcascade_frontalface_default.xml')

# detect faces in the image
face_c = face_cascade.detectMultiScale(img)
face_g = face_cascade.detectMultiScale(gray)

# draw rectangles around faces
for (x, y, h, w) in face_c:
    cv.rectangle(img, (x,y), (x+h, y+w), color = (255, 0, 0), thickness = 2)
for (x, y, h, w) in face_g:
    cv.rectangle(gray, (x,y), (x+h, y+w), color=(255,255,255), thickness = 2)

# display the results
plt.subplot(223)
plt.imshow(img)
plt.subplot(224)
plt.imshow(gray, cmap = 'gray')


plt.show()
```

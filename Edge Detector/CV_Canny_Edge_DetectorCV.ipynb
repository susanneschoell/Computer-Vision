```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt

# load image
image_path = './Panorama.jpg'
# read image
img = cv.imread(image_path)
# convert to grayscale
rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
plt.figure(figsize=(12,8))
plt.imshow(rgb)
plt.axis('off')
plt.figure(figsize=(12,8))
plt.imshow(gray, cmap = 'gray')
plt.axis('off')


# apply the Canny Edge Detector using OpenCV:
edges = cv.Canny(gray,80,600) # threshold values min / max

# The lower threshold value of 80 = surely NOT edges, so ignored / suppressed to zeros
# The upper threshold value of 600 = those are surely edges
# Inbetween are edges or non-edges based on their connectivity = if they are connected to a sure-edge pixel, they are part of an edge
# and vice verca.
# this is to decide which of all detected edges are real edges.

plt.figure(figsize = (12,8))
plt.imshow(edges, cmap = 'gray')
plt.axis('off')

plt.show()
```

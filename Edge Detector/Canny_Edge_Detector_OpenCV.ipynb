```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt

# load image
image_path = '/home/lenovo/anaconda3/jupyter/Python/GeeksforGeeks/OpenCV/lion.jpg'
# read image
img = cv.imread(image_path)
# convert to grayscale
color = cv.cvtColor(img, cv.COLOR_BGR2RGB)
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
plt.figure(figsize=(12,8))
plt.imshow(color)
plt.axis('off')
plt.figure(figsize=(12,8))
plt.imshow(gray, cmap = 'gray')
plt.axis('off')

# give an error message whenever faulty image data occur:
if img is None:
    print('Error, image data faulty')

# simply apply the Canny Edge Detector using OpenCV:
edges = cv.Canny(gray,80,400) # first argument = image, second and third argument = threshold values min / max = Hysteresis Thresholding
# The lower threshold value of 80 = surely NOT edges, so ignored / suppressed to zeros
# The upper threshold value we declared to 400 = those are surely edges
# Inbetween are edges or non-edges based on their connectivity = if they are connected to a sure-edge pixel, they are part of an edge
# and vice verca.
# this is to decide which of all detected edges are real edges.


# Those who lie between these two thresholds are classified edges or non-edges based on their connectivity. If they are connected to
# "sure-edge" pixels, they are considered to be part of edges. Otherwise, they are also discarded.

plt.figure(figsize = (12,8))
plt.imshow(edges, cmap = 'gray')
plt.axis('off')

plt.show()
```
